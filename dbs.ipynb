```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.neighbors import NearestNeighbors
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.neighbors import NearestNeighbors
import plotly.express as px
from sklearn.cluster import KMeans
```

```python
df_scaled = pd.read_csv("dataset_cleaned.csv")
```

```python
#NO pca full features taken for dbscan
MIN_SAMPLES = 46 
neighbors = NearestNeighbors(n_neighbors=MIN_SAMPLES)
neighbors_fit = neighbors.fit(df_scaled)
distances, indices = neighbors_fit.kneighbors(df_scaled)
distances = np.sort(distances[:, MIN_SAMPLES-1], axis=0)
plt.figure(figsize=(10, 6))
plt.plot(distances, color='purple', linewidth=2)
plt.title(f'K-Distance Graph to find optimal eps (k={MIN_SAMPLES})', fontsize=16, fontweight='bold')
plt.xlabel('Customers sorted by distance to 46th neighbor', fontsize=12)
plt.ylabel('Euclidean Distance (eps)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

optimal eps value : 5.5 as seen from the graph

```python
dbscan = DBSCAN(eps=5.5, min_samples=46)
db_labels = dbscan.fit_predict(df_scaled)
df_scaled['DBSCAN_Cluster'] = db_labels
n_clusters = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise = list(db_labels).count(-1)

print("--- DBSCAN RESULTS ---")
print(f"Number of valid clusters: {n_clusters}")
print(f"Number of noise points: {n_noise}")
print("\nCustomer Distribution:")
print(df_scaled['DBSCAN_Cluster'].value_counts().sort_index())
```

```python
#Trying with lower dimensional PCA = 3 componenets
pca = PCA(n_components=3, random_state=42)
pca_data = pca.fit_transform(df_scaled.drop(columns=['DBSCAN_Cluster'], errors='ignore'))
explained_variance = sum(pca.explained_variance_ratio_) * 100
print(f"Variance retained by 3 Principal Components: {explained_variance:.2f}%\n")

NEW_MIN_SAMPLES = 6

neighbors_pca = NearestNeighbors(n_neighbors=NEW_MIN_SAMPLES)
neighbors_fit_pca = neighbors_pca.fit(pca_data)
distances_pca, indices_pca = neighbors_fit_pca.kneighbors(pca_data)

distances_pca = np.sort(distances_pca[:, NEW_MIN_SAMPLES-1], axis=0)

plt.figure(figsize=(10, 6))
plt.plot(distances_pca, color='teal', linewidth=2)
plt.title(f'PCA K-Distance Graph to find optimal eps (k={NEW_MIN_SAMPLES})', fontsize=16, fontweight='bold')
plt.xlabel('Customers sorted by distance to 6th neighbor', fontsize=12)
plt.ylabel('Euclidean Distance (eps)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

optimal eps value : 1.0

```python
pca_dbscan = DBSCAN(eps=1.0, min_samples=6)
pca_db_labels = pca_dbscan.fit_predict(pca_data)
df_scaled['PCA_DBSCAN_Cluster'] = pca_db_labels

n_pca_clusters = len(set(pca_db_labels)) - (1 if -1 in pca_db_labels else 0)
n_pca_noise = list(pca_db_labels).count(-1)

print("--- PCA + DBSCAN RESULTS ---")
print(f"Number of valid clusters: {n_pca_clusters}")
print(f"Number of noise points: {n_pca_noise}")
print("\nCustomer Distribution:")
print(df_scaled['PCA_DBSCAN_Cluster'].value_counts().sort_index())
```

Visualizing why DBSCAN is failing

```python
kmeans_final = KMeans(n_clusters=5, init='k-means++', random_state=42, n_init='auto')
features_for_clustering = df_scaled.drop(columns=['DBSCAN_Cluster', 'PCA_DBSCAN_Cluster'], errors='ignore')
final_labels = kmeans_final.fit_predict(features_for_clustering)
df_pca_viz = pd.DataFrame(pca_data, columns=['PC1', 'PC2', 'PC3'])
df_pca_viz['Cluster'] = final_labels.astype(str) 

fig = px.scatter_3d(
    df_pca_viz, 
    x='PC1', 
    y='PC2', 
    z='PC3',
    color='Cluster', 
    opacity=0.7,
    title='3D Visualization of K-Means (k=5) Segments',
    color_discrete_sequence=px.colors.qualitative.Set1 
)

fig.update_traces(marker=dict(size=4))

fig.show()
```

This is because of the fact that the dataset has consumers very close to each other in the feature space.This presents a significant problem to the DBSCAN algorithm as it tries to fit every thing into one cluster. After applying PCA and reducing dimensions and tuning hyperparameters we could only manage with 2 clusters which is clearly failing in clustering consumers.
